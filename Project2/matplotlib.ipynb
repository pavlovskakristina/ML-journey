```python
# 3rd method (recommended)
fig, ax = plt.subplots()
ax.plot(x, y);
```

# Scater Plot and Bar Plot

```python
# Creating data 
x = np.linspace(0, 10, 100)
x
```

## Line plot

```python
# Ploting data
fig, ax = plt.subplots()
ax.plot(x, x**2);
```

## Scater plot

```python
fig, ax = plt.subplots()
ax.scatter(x, np.exp(x));
```

```python
fig, ax = plt.subplots()
ax.scatter(x, np.sin(x*2));
```

## Bar plot

```python
fig, ax = plt.subplots()
ax.bar(x, x**2);
```

```python
fig, ax = plt.subplots()
ax.bar(x, np.sin(x));
```

```python
# Making a plot from dictionary
butter_price = {"Almond butter": 10,
                "Peanut butter": 8,
                "Cashew butter": 12,
}
fig, ax = plt.subplots()
ax.bar(butter_price.keys(), butter_price.values())
ax.set(title='Butter price',
      ylabel='Price ($)');
```

```python
# barh
fig, ax = plt.subplots()
ax.barh(list(butter_price.keys()), list(butter_price.values()));
```

## Histogram

```python
x = np.random.randn(1000) # Rozkład Gaussa 
fig, ax = plt.subplots()
ax.hist(x);
```

## Subplots

### Subplots option 1

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots( nrows=2,
                                              ncols=2,
                                              figsize=(10,5)
) 
# PLoting each different grids 
ax1.plot(x, x/2);
ax1.plot(x, x/5);

ax2.scatter(np.random.random(10), np.random.random(10));
ax2.scatter(np.random.random(10), np.random.random(10));
ax2.scatter(np.random.random(10), np.random.random(10));

ax3.bar(butter_price.keys(), butter_price.values());

ax4.hist(np.random.randn(1000));
```

### Subplots option 2

```python
fig, ax = plt.subplots( nrows=2,
                        ncols=2,
                        figsize=(10,5)
) 
ax[0, 0].plot(x, x/2);
ax[0, 0].plot(x, x/5);

ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[0, 1].scatter(np.random.random(10), np.random.random(10));

ax[1, 0].bar(butter_price.keys(), butter_price.values());

ax[1, 1].hist(np.random.randn(1000));
```
