```python
# Making a plot from dictionary
butter_price = {"Almond butter": 10,
                "Peanut butter": 8,
                "Cashew butter": 12,
}
fig, ax = plt.subplots()
ax.bar(butter_price.keys(), butter_price.values())
ax.set(title='Butter price',
      ylabel='Price ($)');
```

```python
# barh
fig, ax = plt.subplots()
ax.barh(list(butter_price.keys()), list(butter_price.values()));
```

## Subplots

### Subplots option 1

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots( nrows=2,
                                              ncols=2,
                                              figsize=(10,5)
) 
# PLoting each different grids 
ax1.plot(x, x/2);
ax1.plot(x, x/5);

ax2.scatter(np.random.random(10), np.random.random(10));
ax2.scatter(np.random.random(10), np.random.random(10));
ax2.scatter(np.random.random(10), np.random.random(10));

ax3.bar(butter_price.keys(), butter_price.values());

ax4.hist(np.random.randn(1000));
```

### Subplots option 2

```python
fig, ax = plt.subplots( nrows=2,
                        ncols=2,
                        figsize=(10,5)
) 
ax[0, 0].plot(x, x/2);
ax[0, 0].plot(x, x/5);

ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[0, 1].scatter(np.random.random(10), np.random.random(10));
ax[0, 1].scatter(np.random.random(10), np.random.random(10));

ax[1, 0].bar(butter_price.keys(), butter_price.values());

ax[1, 1].hist(np.random.randn(1000));
```

# Plotting from Pandas dataframes

```python
car_sales = pd.read_csv("auta_treningowe.csv")
car_sales.head(11)
```

```python
print(type(car_sales['Cena_PLN']))
type(car_sales)
```

```python
# We are creating here Sales data column
car_sales["Data sprzedaży"] = pd.date_range("1/1/2025", periods=len(car_sales))
print("Długość listy:", len(car_sales))
```

```python
car_sales
```

## CUMSUM - funkcja sumuje wszystkie wierze w kolumnie ze soba 

```python
car_sales["ZYSK CAŁKOWITY"] = car_sales["Cena_PLN"].cumsum()
car_sales.head(10)
```

```python
# Plotting total sales 

car_sales.plot(x="Data sprzedaży", y="ZYSK CAŁKOWITY");
```

```python
car_sales["ZYSK CAŁKOWITY"] = car_sales["ZYSK CAŁKOWITY"]/10000
car_sales.plot(x="Data sprzedaży", y="ZYSK CAŁKOWITY");
```

```python
import matplotlib.pyplot as plt
x = car_sales["Przebieg_km"]
y = car_sales["Cena_PLN"]
plt.scatter(x, y);
title = "Wpływ przebiegu na cenę [PLN]"
plt.title(title)
plt.xlabel("Przebieg auta [km]")
plt.ylabel("Cena auta [PLN]")
plt.show();


```
