```python
# importing libraries
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
data = pd.read_csv("auta_treningowe.csv")
data.head(10)
```

```python
pd.crosstab(data["Marka"], data["Bezwypadkowy"])
```

```python
# Groupby - grupuje dane według kolumny "Marka". 
# Czyli wszystkie wiersze z tą samą marką trafiają do jednej grupy
# .mean() - dla każdej grupy liczy średnią ze wszystkich kolumn

data.groupby(["Marka"]).mean(numeric_only=True)
```

```python
# czasami się używa % - magiczna sztuczka

%matplotlib inline
data["Cena_PLN"].plot()
```

```python
srednia_cena = data.groupby("Marka")["Cena_PLN"].mean()
srednia_cena.plot()
```

```python
# Rodzaj paliwa
pd.crosstab(data["Marka"], data["Rodzaj_paliwa"]).plot(kind="bar", stacked=True)
```

```python
# hist() - pozwaa nam zobaczyć dystrybucję - rozprzestrzenienie się danych 
data["Rok_produkcji"].hist() 
```

```python
data["Cena_PLN"].hist() 
```
