# Heart disease

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
heart_disease = pd.read_csv("heart-disease.csv")
print("Number of patients: ", len(heart_disease))
heart_disease.head(5)
```

```python
# Histogram  of age

ax = heart_disease["age"].plot.hist(bins=50)
ax.set_title("Najczęstszy wiek zawału (bez podziału na płeć)")
ax.set_xlabel("Wiek")
ax.set_ylabel("Pacjenci")

plt.show()
```

```python
heart_disease.head(5)
```

```python
# Histogram of age

woman = heart_disease[heart_disease["sex"] == 0]
woman.head()
```

```python
# MOŻNA TEŻ TAK GRUPOWAĆ
grouped = heart_disease.groupby("sex")
woman = grouped.get_group(0)
woman.head()
```

```python
woman["age"].plot.hist(bins=50)
plt.title("Najczęstszy wiek zawału (dla kobiet)")
plt.xlabel("Wiek")
plt.ylabel("Liczba kobiet")

plt.show()
```

```python
print(f"Zawał u kobiet wystpił w {len(woman)} przypadkach")
print(f"Zawał u mężczyzn wystpił w {len(heart_disease) - len(woman)} przypadkach")
print("Łaczna liczba zawałów na serce:", len(heart_disease))
```

```python
over_50 = heart_disease[heart_disease["age"] > 50]
over_50.head()
```

```python
## Pyplot method
over_50.plot(kind="scatter",
            x="age",
            y="chol",
            c="target",
            figsize=(15, 11));
```

```python
## OO method

fix, ax = plt.subplots(figsize=(10,8))
over_50.plot(
    kind="scatter",
    x="age",
    y="chol",
    c="target",
    cmap="winter",  ### COLOR (also: winter, summer, plasma)
    ax=ax
);

# matplotlib.tutorials/colors

ax.set_xlim([45, 80])
```

## OO method from scratch 

```python
fig, ax = plt.subplots(figsize=(10, 8))

# Plotting the data
scatter = ax.scatter(
    x=over_50["age"],
    y=over_50["chol"],
    c=over_50["target"],
)

# Customize the plot

ax.set(
    title="Heart Didease and Cholesterol Levels",
    xlabel="Age",
    ylabel="Cholesterol"
);

# Adding legend
ax.legend(*scatter.legend_elements(), title="Target");

# Adding a horizontal line
ax.axhline(over_50["chol"].mean(),
          linestyle="--");
```

```python
# Subplot of chol, age, thalach 
```

```python
# Plotting 

fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10)
                              )

# Adding data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                     )

# Customizing ax0
ax0.set(title="Heart Disease and Cholesterol Level",
        xlabel="Age",
        ylabel="CHolesterol"
)

# Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target")

# Adding a meanline
ax0.axhline(y=over_50["chol"].mean(),
            linestyle="--"
           )
## SECOND PLOT

# Adding data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"],
                     )

# Customizing ax1
ax1.set(title="Heart Disease and Max Heart Rate",
        xlabel="Age",
        ylabel="Thalach"
)

# Add a legend to ax1
ax1.legend(*scatter.legend_elements(), title="Target")

# Adding a meanline
ax1.axhline(y=over_50["thalach"].mean(),
            linestyle="--"
           )



    
plt.show();
```

```python
# THE SAME BUT WITH SHARING  #sharex= True

# Plotting 

fig, (ax0, ax1) = plt.subplots(nrows=2,
                              ncols=1,
                              figsize=(10, 10),
                              sharex= True  # HERE 
                              )

# Adding data to ax0
scatter = ax0.scatter(x=over_50["age"],
                      y=over_50["chol"],
                      c=over_50["target"],
                     )

# Customizing ax0
ax0.set(title="Heart Disease and Cholesterol Level",
        # xlabel="Age", REMOVED
        ylabel="CHolesterol"
)

# Add a legend to ax0
ax0.legend(*scatter.legend_elements(), title="Target")

# Adding a meanline
ax0.axhline(y=over_50["chol"].mean(),
            linestyle="--"
           )
## SECOND PLOT

# Adding data to ax1
scatter = ax1.scatter(x=over_50["age"],
                      y=over_50["thalach"],
                      c=over_50["target"],
                     )

# Customizing ax1
ax1.set(title="Heart Disease and Max Heart Rate",
        xlabel="Age", 
        ylabel="Thalach"
)

# Add a legend to ax1
ax1.legend(*scatter.legend_elements(), title="Target")

# Adding a meanline
ax1.axhline(y=over_50["thalach"].mean(),
            linestyle="--"
           )

# Adding a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");

    
plt.show();
```
