## Wariancja (variance)

Miara statystyczna, która określa, jak bardzo dane są rozproszone (rozrzucone) wokół ich średniej arytmetycznej. Im wyższa wariancja, tym bardziej wartości odstają od średniej. 
Wariancja jest nam potrzebna, aby zrozumieć stopień niepewności, szumu lub zmienności w analizowanym systemie.

1. Wykrywanie "szumu" i stabilności sieci (Telekomunikacja)
Wyobraź sobie, że analizujesz opóźnienia (ping) na dwóch różnych serwerach. Oba mają średni ping na poziomie 50 ms.

        Serwer A: 49 ms, 50 ms, 51 ms (Wariancja bliska zeru — połączenie jest niezwykle stabilne).

        Serwer B: 10 ms, 50 ms, 90 ms (Bardzo wysoka wariancja — połączenie szarpie).

## Odchylenie standardowe (standard deviation)

Najbardziej intuicyjna i najczęściej stosowana miara tego, jak bardzo dane są rozproszone wokół średniej. Dzięki temu odchylenie mówi Ci wprost: "Twoje dane odchylają się od średniej przeciętnie o X jednostek".

W analizie danych i uczeniu maszynowym bardzo często zakłada się, że dane mają tzw. rozkład normalny (wykres przypominający dzwon). Dla takich danych odchylenie standardowe pozwala natychmiast ocenić, co jest "normą", a co "anomalią":

    ~68% wszystkich Twoich danych znajduje się w odległości 1 odchylenia standardowego od średniej.

    ~95% danych mieści się w odległości 2 odchyleń.

    ~99.7% danych mieści się w odległości 3 odchyleń (wszystko poza tym to bardzo rzadkie anomalie, tzw. outliery).

Przykład: Jeśli średni wzrost w grupie to 170 cm, a odchylenie standardowe to 10 cm, to wiesz, że około 68% osób z tej grupy ma wzrost między 160 cm a 180 cm.

## Variance and standard deviation

```python
higher_var_array = np.array([1, 100, 200, 300, 4000, 5000])
low_var_array = np.array([2, 4, 6, 8, 10])
```

```python
print("Variance")
print(f"High: {np.var(higher_var_array)} ")
print(f"Low: {np.var(low_var_array)} ")
```

```python
print("Standard deviation")
print(f"High: {np.std(higher_var_array)} ")
print(f"Low: {np.std(low_var_array)} ")
```

```python
np.mean(higher_var_array), np.mean(low_var_array)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.hist(higher_var_array)
plt.show()
```

```python
plt.hist(low_var_array)
plt.show()
```
