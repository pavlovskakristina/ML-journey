# Dot Product

Dot product (czyli iloczyn skalarny) to jedna z najważniejszych operacji matematycznych w całej algebrze liniowej i absolutny fundament sztucznej inteligencji.

W najprostszym ujęciu jest to operacja, która bierze dwa ciągi liczb (wektory) o tej samej długości, mnoży przez siebie odpowiadające im elementy, a na koniec sumuje te wyniki w jedną pojedynczą liczbę (skalar).

$$a \cdot b = (a_1 \times b_1) + (a_2 \times b_2) + \dots + (a_n \times b_n)$$

## Dlaczego potrzebowaliśmy "transpose"?

Kiedy przechodzimy z pojedynczych wektorów na całe macierze (tablice 2D), dot product staje się mnożeniem macierzy.

Zasada jest tutaj ściśle określona: mnożysz wiersze pierwszej macierzy przez kolumny drugiej macierzy. Aby to było fizycznie możliwe, liczba elementów w wierszu pierwszej macierzy musi być identyczna z liczbą elementów w kolumnie drugiej macierzy.Mówiąc językiem kształtów (wymiarów): macierz o wymiarach $M \times N$ można pomnożyć przez dot product tylko z macierzą o wymiarach $N \times P$. 

Wewnętrzne wymiary ($N$) muszą się zgadzać.

Wynikiem będzie nowa macierz o wymiarach $M \times P$.

```python
import numpy as np

# Nasze wektory opisują 3 cechy: [Akcja, Komedia, Sci-Fi]

# Użytkownik bardzo lubi Akcję (5) i Sci-Fi (4), ale nie znosi Komedii (0)
uzytkownik = np.array([5, 0, 4])

# Profilujemy nasze filmy (1 oznacza, że film ma tę cechę, 0 że nie ma)
matrix = np.array([1, 0, 1])         # Akcja + Sci-Fi
kac_vegas = np.array([0, 1, 0])      # Tylko Komedia
men_in_black = np.array([1, 1, 1])   # Akcja + Komedia + Sci-Fi

# Obliczamy podobieństwo używając dot product
wynik_matrix = np.dot(uzytkownik, matrix)
wynik_kac_vegas = uzytkownik @ kac_vegas        # używamy operatora @ (to samo co np.dot)
wynik_men_in_black = uzytkownik @ men_in_black

print(f"Dopasowanie dla Matrixa: {wynik_matrix}")
print(f"Dopasowanie dla Kac Vegas: {wynik_kac_vegas}")
print(f"Dopasowanie dla Men in Black: {wynik_men_in_black}")
```
